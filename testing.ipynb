```python
import joblib
from tensorflow.keras.models import load_model

iso_model = joblib.load("models/isolation_forest.pkl")
xgb_model = joblib.load("models/xgboost_model.pkl")
scaler = joblib.load("models/scaler.pkl")

lstm_autoencoder = load_model(
    "models/lstm_autoencoder.keras"
)

print("✅ Isolation Forest loaded")
print("✅ XGBoost loaded")
print("✅ LSTM Autoencoder loaded")
print("✅ Scaler loaded")
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("data/cleaned_CICIDS2017_EDA.csv")

df.columns = df.columns.str.strip()

X = df.drop(columns=["Label", "Traffic_Type"])

X = X.select_dtypes(include=np.number)

X = X.replace([np.inf, -np.inf], np.nan)

X = X.dropna()

# Keep exactly the features used when training the scaler
X = X[scaler.feature_names_in_]

print("Real traffic shape:", X.shape)
print("Features match: ✅")
```

```python
sample = X.iloc[[0]]

sample_scaled = scaler.transform(sample)

iso_result = iso_model.predict(sample_scaled)[0]

xgb_result = xgb_model.predict(sample)[0]

print("Isolation Forest:", "ANOMALY" if iso_result == -1 else "NORMAL")
print("XGBoost:", "ATTACK" if xgb_result == 1 else "NORMAL")
```
